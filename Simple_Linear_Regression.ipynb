```python
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
x=[[1],[2],[3],[4],[5]]
y=[[3],[6],[9],[12],[15]]
model=LinearRegression()
model.fit(x,y)
print(model.predict([[15]]))
```

```python
#Calculate m and c
m=model.coef_[0]
c=model.intercept_
print('Slope: ',m)
print('Intercept: ',c)
print('Prediction: ',model.predict([[12]]))
```
